```python
import numpy as np
import pandas as pd

x = np.array([18, 42, 66, 90])
x_norm = (x - x.min()) / (x.max() - x.min())
#x_norm = (x - 18) / (90 - 18)

x_norm = np.round(x_norm, 2)
data = {
    'list': [x],
    'list_norm': [x_norm],
}
df = pd.DataFrame(data)
df
```

```python
list = np.array([8, 10, 10, 14, 14, 16])
mean_list = sum(list)/len(list)

#std_List =np.std(list)
std = np.sqrt(sum((list - mean_list)**2)*(1/len(list)))

print(f'mean_list: {mean_list}')
print(f'std: {std}')

scaler_list = (list - mean_list)/std
print(scaler_list)

print()
scaler_list = np.round(scaler_list, 3)
data = {
    'list': [list],
    'list_norm': [scaler_list],
}
df = pd.DataFrame(data)
df
```
